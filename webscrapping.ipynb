how to do webscrapping

```python
# It used for open website or send them to https requests or download pages
# Without this pyhton cannot open websites
import requests 

# It used to read html, find tags, Extract text, links, images
from bs4 import BeautifulSoup

# url is a variable and https is website where we want to scrap
url = "https://www.moneycontrol.com/news/business/markets/"
# headers is variable {} is to store key 
headers = {
    # http header pretends to be real browser without it block the python  
    "User-Agent": "Mozilla/5.0" 
}

#request=HTTp Get request headers=sends browser identity along request
response = requests.get(url, headers=headers) 

#contains html code of webpage parser tells beautifulsoup this content is html
soup = BeautifulSoup(response.text, "html.parser") 

#find all means find all matching tag and h2 means for new titles
headlines = soup.find_all("h2")

#for is loop news is one headline at a time and h2 means go through every new headlines one by one
for news in headlines:
    #news.text means extract only readable html remove html tags strip means extra space and new line
    print(news.text.strip())
```
