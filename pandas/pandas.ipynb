Series

series = pandas 1-Dimensional labeled array that can hold any data type 

Think of it like a single column in a spreadsheet (1-Dimensional) 

```python
data = [1, 2, 3.5, 4.2, 5] # this shows a simple pandas series
series = pd.Series(data, index=['a', 'b', 'c', 'd', 'e']) # this shows a pandas series with custom index
print(series) # it also shows the datatype of the series
```

```python
data = [100,102,103,104,105]
series = pd.Series(data, index=['a', 'b', 'c', 'd', 'e']) # this shows a pandas series with custom index
series.loc ['e'] = 9 # changing value at index 'e' to 9

print(series) # it also shows the datatype of the series

print(series.iloc[4]) # accessing value at index position 4

print(series[series >= 104]) # filtering values greater than or equal to 104

print(series.loc['e']) # it also shows the datatype of the series
```

```python
calories = {'day1': 420, 'day2': 380, 'day3': 390} # creating a dictionary

series = pd.Series(calories)
series.loc['day1'] += 80  # updating value at index 'day1'

print(series)
```

```python
data = {
    "name": ['manav', 'kathan', 'mohit', 'zoro'],
    "age": [24, 23, 23, 32]
}
df = pd.DataFrame(data, index=["partner 1-", "partner 2-", "partner 3-", "partner 4-"]) # creating a dataframe with custom index

print(df.loc["partner 1-"])  # accessing row by index label

print(df.iloc[1])  # accessing row by index position

# add a new column 'partner-5' to the dataframe
df["job"] = ["developer", "developer", "financer", "analyst"]

# add a new row to the dataframe
new_row = pd.DataFrame({"name": "luffy", "age": 29, "job": "manager"}, index=["partner 5-"])
df = pd.concat([df, new_row]) # concatenating the new row to the existing dataframe
print(df)
```
