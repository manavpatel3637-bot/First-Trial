```python
from curl_cffi import requests
import pandas as pd
import time
import os
from datetime import datetime

# Create session with browser impersonation
session = requests.Session(impersonate="chrome")

headers = {
    "Referer": "https://www.nseindia.com/market-data/most-active"
}

# First request to set cookies
session.get("https://www.nseindia.com", headers=headers)
time.sleep(2)

URL = "https://www.nseindia.com/api/live-analysis-most-active-underlying"

while True:
    try:
        response = session.get(URL, headers=headers)
        response.raise_for_status()

        data = response.json()

        # Convert to DataFrame (ALL rows)
        df = pd.DataFrame(data["data"])

        # Clear screen (Windows / Linux / Mac)
        os.system("cls" if os.name == "nt" else "clear")

        print("📊 NSE MOST ACTIVE UNDERLYING")
        print("⏰ Last Updated:", datetime.now().strftime("%Y-%m-%d %H:%M:%S"))
        print("-" * 80)

        # Show ALL data
        pd.set_option("display.max_rows", None)
        pd.set_option("display.max_columns", None)
        pd.set_option("display.width", None)

        print(df)

    except Exception as e:
        print("❌ Error:", e)

    # Refresh every 5 minutes
    print("\n🔄 Refreshing again in 5 minutes...\n")
    time.sleep(300)

high_volume = df[df["totalTradedVolume"] > 1_00_00_000]
print(high_volume)
```
